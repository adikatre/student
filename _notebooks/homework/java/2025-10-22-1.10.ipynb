```python
public class UltimateBattle {

    public static void main(String[] args) {
        // Create two fighters
        Fighter robot = new Fighter("RoboMax", 25, 100);
        Fighter dinosaur = new Fighter("T-Rex", 30, 100);

        // Print initial status
        robot.printStatus();
        dinosaur.printStatus();

        // Use instance methods to attack
        robot.attack(dinosaur);
        dinosaur.attack(robot);

        // Print status after attacks
        robot.printStatus();
        dinosaur.printStatus();

        // Use static methods
        Fighter.strongerFighter(robot, dinosaur);
        Fighter.beginBattle(robot, dinosaur);
    }
}

class Fighter {
    // Instance variables
    String name;
    int power;
    int health;

    // Static variable
    static double fightDuration = 3.5; // in minutes

    // Constructor
    public Fighter(String name, int power, int health) {
        this.name = name;
        this.power = power;
        this.health = health;
    }

    // Instance method 1: Attack
    void attack(Fighter opponent) {
        System.out.println(this.name + " attacks " + opponent.name + " for " + power + " damage!");
        opponent.health -= power;
        if (opponent.health < 0) opponent.health = 0;
    }

    // Instance method 2: Print status
    void printStatus() {
        System.out.println(name + " → Health: " + health + " | Power: " + power);
    }

    // Static method 1: Compare fighters
    static void strongerFighter(Fighter a, Fighter b) {
        if (a.power > b.power)
            System.out.println(a.name + " is the stronger fighter!");
        else if (a.power < b.power)
            System.out.println(b.name + " is the stronger fighter!");
        else
            System.out.println("Both fighters have equal power!");
    }

    // Static method 2: Begin battle
    static void beginBattle(Fighter a, Fighter b) {
        System.out.println("\n⚔️ The battle between " + a.name + " and " + b.name + " begins!");
        System.out.println("Estimated duration: " + fightDuration + " minutes.\n");
    }
}
```
