```java
// ===========================
// Popcorn Hack 2.1: Input Validation
// ===========================

import java.util.InputMismatchException;
import java.util.Scanner;

public class InputValidationDemo {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);

        System.out.print("Enter a number: ");
        try {
            int num = sc.nextInt();
            System.out.println("You entered: " + num);
        } catch (InputMismatchException e) {
            System.out.println("Error: You must enter a valid integer!");
        }

        sc.close();
    }
}

// Answer:
// If the user types "twenty", the program throws InputMismatchException.
// Handling exceptions makes the program more robust and user-friendly.

InputValidationDemo.main(null);
```

```java
// ===========================
// Popcorn Hack 2.2: Token vs Line
// ===========================

import java.util.Scanner;

public class TokenVsLine {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);

        System.out.print("Enter city name with spaces: ");
        String token = sc.next();        // Reads one word (up to whitespace)
        sc.nextLine();                   // Clear buffer
        String line = sc.nextLine();     // Reads the full line

        System.out.println("With next(): " + token);
        System.out.println("With nextLine(): " + line);

        sc.close();
    }
}

// Answer:
// If input is "San Diego":
// next() -> "San"
// nextLine() -> "San Diego"

TokenVsLine.main(null);
```

```java
// ===========================
// Homework Hack 1: Three-Number Average Calculator
// ===========================

import java.util.Scanner;

public class AverageCalculator {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);

        System.out.print("Enter first number: ");
        int n1 = sc.nextInt();
        System.out.print("Enter second number: ");
        int n2 = sc.nextInt();
        System.out.print("Enter third number: ");
        int n3 = sc.nextInt();

        double average = (n1 + n2 + n3) / 3.0; // Cast to double for decimal result
        System.out.printf("Average: %.2f%n", average);

        sc.close();
    }
}

AverageCalculator.main(null)
```

```java
// ===========================
// Homework Hack 2: Assignment vs Comparison
// ===========================

// Assignment Operator (=)
// - Stores a value in a variable
// Example:
int a = 5;  // Assigns 5 into variable a

// Comparison Operator (==)
// - Checks if two values are equal (returns true/false)
// Example:
boolean result = (a == 5);   // true

// Common Bug Example:
if (a = 5) { ... }   // ERROR: trying to assign instead of compare
// Should be:
if (a == 5) { ... }  // Checks equality

// Explanation:
// '=' changes the variable, '==' compares values.
```
